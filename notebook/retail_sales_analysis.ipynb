```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

print("Environment Ready 🚀")
```

```python
df = pd.read_csv("../data/SuperStoreOrders.csv")
df.head()
```

```python
df.info()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='mixed')
```

```python
df.info()
```

```python
df['sales'].head()
```

```python
df['sales'] = df['sales'].str.replace(',', '', regex=False)
```

```python
df['sales'] = df['sales'].astype(float)
```

```python
df.info()
```

```python
sales_per_year = df.groupby('year')['sales'].sum().reset_index()
sales_per_year
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(sales_per_year['year'], sales_per_year['sales'])
plt.title('Total Sales Per Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure()
plt.plot(sales_per_year['year'], sales_per_year['sales'], marker='o')
plt.title('Total Sales per year')
plt.xlabel('year')
plt.ylabel('Total Sales')
plt.show()
```

```python
profit_per_year = df.groupby('year')['profit'].sum().reset_index()
```

```python
plt.figure()
plt.plot(profit_per_year['year'], profit_per_year['profit'], marker='o')
plt.title('Total Profit per year')
plt.xlabel('Year')
plt.ylabel('Total Profit')
plt.show()
```
